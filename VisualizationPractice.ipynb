```python
from matplotlib import pyplot as plt
```

```python
age= [25,26,27,28,29,30,31,32,33]
```

```python
tata = [2000,3000,5126,7823,2000,3012,6720,8000,4000]
```

```python
infi=[3000,2567,5678,8920,8930,4563,2400,6000,5340]
```

```python
plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Agewise-Salary distribution")

plt.plot(age,tata,"b--",marker='.',label='Tata')
plt.plot(age,infi,color='y',linestyle='--',label="Infi",marker='o',linewidth=3)
plt.legend()
plt.savefig("sal_age.jpg")
#plt.grid()
plt.style.use('fivethirtyeight')
```

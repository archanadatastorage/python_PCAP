```python
#Data Cleaning - Handling Null Values
#Create Data with Null Values


# Create data with missing values
data_with_nulls = """Name,Age,City,Salary,Department
John,28,New York,50000,IT
Alice,,Los Angeles,60000,HR
Bob,45,,55000,IT
Carol,29,Houston,,Finance
David,38,Phoenix,58000,
Emma,,Philadelphia,54000,IT
Frank,42,San Antonio,,Finance
Grace,26,San Diego,51000,Marketing
Henry,,,56000,IT
Ivy,30,San Jose,53000,HR"""

with open('employees_nulls.csv', 'w') as f:
    f.write(data_with_nulls)

df = pd.read_csv('employees_nulls.csv')
print(df)
```

```python
#Check for Null Values

# Count null values in each column
print(df.isnull().sum())

# Count total null values
print(f"Total null values: {df.isnull().sum().sum()}")

# Show rows with any null values
print(df[df.isnull().any(axis=1)])

# Check specific column
print(df['Age'].isnull().sum())
```

```python
#Remove Null Values

# Remove rows with ANY null values
df_cleaned = df.dropna()
print(df_cleaned)

# Remove rows where ALL values are null
df_cleaned = df.dropna(how='all')

# Remove rows with nulls in specific columns
df_cleaned = df.dropna(subset=['Age', 'Salary'])

# Remove columns with null values
df_cleaned = df.dropna(axis=1)
```

```python
#Replace Null Values

# Replace ALL null values with a specific value
df_filled = df.fillna(0)
print(df_filled)

# Replace nulls with different values per column
df_filled = df.fillna({
    'Age': 0,
    'City': 'Unknown',
    'Salary': df['Salary'].mean(),  # Use mean salary
    'Department': 'Not Assigned'
})
print(df_filled)

# Forward fill (use previous value)
df_filled = df.fillna(method='ffill')

# Backward fill (use next value)
df_filled = df.fillna(method='bfill')

# Replace with mean (numeric columns only)
df['Age'].ffill(df['Age'].mean(), inplace=True)

# Replace with median
df['Salary'].fillna(df['Salary'].median(), inplace=True)

# Replace with mode (most common value)
df['Department'].fillna(df['Department'].mode()[0], inplace=True)
```
